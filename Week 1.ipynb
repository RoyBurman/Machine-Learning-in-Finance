```python
# Machine Learning in Finance 
# Week 1
# My SID : 2546621   'Last 2 digits: 21', we'll use 21 elements

import numpy as np

# Making a Maxtrix with 21 Elements
array = np.arange(21)
print("Original Matrix :")
print(array)

# Changing the 2D Matrix with 1 Row
array = array.reshape(1, -1)
print("\n 2D array with 1 row : ")
print(array)

# Saving the original array to a different array
diff_array = array
print("\n Coppied Array :")
print(diff_array)

# Shape of the diff_array ?
print("\n Shape of the diff_array")
print(diff_array.shape)
```
